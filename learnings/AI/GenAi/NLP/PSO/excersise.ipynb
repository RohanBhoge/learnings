```python
import spacy
```

```python
with open("news.txt", encoding="utf-8") as f:
    text = f.read()
```

```python
nlp=spacy.load("en_core_web_sm")
doc=nlp(text)
```

```python
for token in doc:
    if(token.pos_ == "NOUN"):
        print(token.text, " | ", token.pos_," | ", spacy.explain(token.pos_))
```

```python
for token in doc:
    if(token.pos_ == "NUM"):
        print(token.text, " | ", token.pos_," | ", spacy.explain(token.pos_))
```

```python
from collections import Counter

# Count POS tags in the doc
pos_counts = Counter(token.pos_ for token in doc)

# Print total number of POS tags
print(f"Total POS tags in the file: {sum(pos_counts.values())}")
print("POS tag counts:")
for pos, count in pos_counts.items():
    print(f"{pos}: {count}")
```

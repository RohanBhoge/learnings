```python
from langchain_google_genai import ChatGoogleGenerativeAI  # imported class 
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash-exp")

# Use the LLM
response = llm.invoke("Hi, how are you?")
print(response.content)
```

```python
from langchain.prompts import PromptTemplate
Prompt_Template_name=PromptTemplate(
    input_variables=["cusine"],
    template="I want to open restorent for {cusine} food. suggest fancey name for restorent.",
)

Prompt_Template_name.format(cusine="Mexican food")
```

```python
from langchain.chains import LLMChain
from langchain.memory import ConversationBufferMemory

memory = ConversationBufferMemory()

chain = LLMChain(llm=llm, prompt=Prompt_Template_name,memory=memory)
chain.run("Indian restorent")
```

```python
print(chain.memory.buffer)
```

```python
from langchain.chains import ConversationChain
convo=ConversationChain(llm=llm, memory=memory)
print(convo.prompt.template)
```

```python
convo.run("suggest me some brain teasers")
```

```python
print(convo.prompt.template)
print(convo.memory.buffer)
```

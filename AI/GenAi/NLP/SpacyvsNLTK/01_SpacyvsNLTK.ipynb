```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm")
doc=nlp("The quick brown fox jumped over the lazy dog. Then dog is goes crazy.")
for sentence in doc.sents:
    print(sentence)

    # Print each word in the sentence
    for word in sentence:
        print(word)
```

```python
import nltk

import os

from nltk.tokenize import sent_tokenize, word_tokenize

# Set the NLTK data path explicitly
nltk_data_path = os.path.join(os.path.expanduser('~'), 'AppData', 'Roaming', 'nltk_data')
nltk.data.path.append(nltk_data_path)

# Verify the directory exists, create if it doesn't
if not os.path.exists(nltk_data_path):
    os.makedirs(nltk_data_path)

# Retry the download, specifying the download directory
try:
    nltk.download('punkt', download_dir=nltk_data_path)
except Exception as e:
    print(f"Error downloading punkt: {e}")
    print("Please check your internet connection and write permissions to the NLTK data directory.")
    print("You may also try manually downloading punkt from the NLTK website and placing it in the NLTK data directory.")

sentToken=sent_tokenize("The quick brown fox jumped over the lazy dog. Then dog is goes crazy.")
wordToken=word_tokenize("The quick brown fox jumped over the lazy dog. Then dog is goes crazy.")

```

```python
for token in wordToken:
    print(token)
```
